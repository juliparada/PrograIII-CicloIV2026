```python
while True:
    sueldo = float(input("Ingrese el sueldo: $"))

    if sueldo * 0.03 > 30.00:
        isss = 30.00
    else:
        isss = sueldo * 0.03

    if sueldo * 0.0725 > 507.50:
        afp = 507.50
    else:
        afp = sueldo * 0.0725

    renta_gravada = sueldo - isss - afp

    if renta_gravada <= 550.00:
        isr = 0.00
    elif renta_gravada <= 895.24:
        isr = (17.67 + (renta_gravada - 550.00) * 0.10)
    elif renta_gravada <= 2038.10:
        isr = (60.00 + (renta_gravada - 895.24) * 0.20)
    else:
        isr = (288.57 + (renta_gravada - 2038.10) * 0.30)

    total_deducciones = isss + afp + isr

    sueldo_neto = sueldo - total_deducciones

    print("\n--- DEDUCCIONES ---")
    print(f"Sueldo:             ${sueldo:.2f}")
    print(f"ISSS (3%):          ${isss:.2f}")
    print(f"AFP (7.25%):        ${afp:.2f}")
    print(f"ISR:                ${isr:.2f}")
    print(f"Total deducciones:  ${total_deducciones:.2f}")
    print(f"Sueldo neto:        ${sueldo_neto:.2f}")

    continuar = input("\n¿Desea calcular otro sueldo? (s/n): ").lower()
    if continuar != 's':
        break
print("Fin del cálculo de sueldo.")
```

6.Nivel de prioridad de pacientes

```python
import datetime

def evaluar_prioridad(edad, motivo):
    motivo_lower = motivo.lower()

    # PRIORIDAD 1: Emergencias vitales o signos de gravedad agudos
    rojo_directo = [
        "inconsciente", "paro", "shock", "choque", "hemorragia masiva",
        "politraumatismo", "no respira", "convulsion", "paro cardiaco", "infarto"
    ]
    if any(kw in motivo_lower for kw in rojo_directo):
        return "Prioridad 1"

    sospecha_emergencia = ["dolor de pecho", "dificultad para respirar", "falta de aire", "asfixia"]
    if any(kw in motivo_lower for kw in sospecha_emergencia):
        resp = input("  [?] ¿El paciente presenta dificultad respiratoria severa o dolor de pecho agudo? (s/n): ").strip().lower()
        if resp == 's':
            return "Prioridad 1"

    if edad < 5 and any(kw in motivo_lower for kw in ["fiebre", "tos", "diarrea", "vomito", "grave"]):
        resp = input("  [?] ¿El menor presenta letargia, convulsiones o incapacidad para beber/beber pecho? (s/n): ").strip().lower()
        if resp == 's':
            return "Prioridad 1"

    if any(kw in motivo_lower for kw in ["embarazo", "embarazada", "parto", "puerpera", "sangrado vaginal"]):
        resp = input("  [?] ¿Presenta sangrado vaginal, convulsiones o fuerte dolor abdominal? (s/n): ").strip().lower()
        if resp == 's':
            return "Prioridad 1"

    # PRIORIDAD 2: Vulnerables o casos agudos/descompensados
    if edad >= 65 or edad < 1:
        return "Prioridad 2"

    if any(kw in motivo_lower for kw in ["embarazo", "embarazada", "puerpera"]):
        return "Prioridad 2"

    p2_keywords = [
        "descompensad", "diabetes", "hipertension", "presion alta", "presion baja",
        "dolor fuerte", "dolor moderado", "fractura", "quemadura", "alergia", "asma"
    ]
    if any(kw in motivo_lower for kw in p2_keywords):
        return "Prioridad 2"

    # PRIORIDAD 3: Consultas leves o rutinarias
    return "Prioridad 3"


def sistema_triage():
    pacientes = []
    orden_prioridad = {"Prioridad 1": 1, "Prioridad 2": 2, "Prioridad 3": 3}
    contador_llegada = 0  # Garantiza el orden de llegada exacto

    while True:
        print("\n--- MENÚ PRINCIPAL ---")
        print("1. Agregar paciente")
        print("2. Ver lista de pacientes por orden de atención")
        print("3. Salir")
        opcion = input("Seleccione una opción: ").strip()

        if opcion == '1':
            print("\n--- NUEVO PACIENTE ---")
            nombre = input("Nombre del paciente: ").strip()

            while True:
                try:
                    edad = int(input("Edad: "))
                    if edad >= 0:
                        break
                    print("Ingrese una edad válida.")
                except ValueError:
                    print("Por favor, ingrese un número entero.")

            motivo = input("Motivo de consulta: ").strip()

            prioridad = evaluar_prioridad(edad, motivo)
            contador_llegada += 1  # Asigna turno correlativo según llegada

            pacientes.append({
                "nombre": nombre,
                "motivo": motivo,
                "prioridad": prioridad,
                "orden": orden_prioridad[prioridad],
                "llegada": contador_llegada
            })

            print(f"\n[OK] Paciente {nombre.upper()} asignado a: {prioridad}")

            sub_opcion = input("\n¿Desea agregar otro paciente? (s/n): ").strip().lower()
            if sub_opcion != 's':
                pass

        elif opcion == '2':
            print("\n--- LISTA DE PACIENTES (ORDEN DE ATENCIÓN) ---")
            if not pacientes:
                print("No hay pacientes registrados.")
            else:
                # Se ordena PRIMERO por 'orden' (Prioridad 1, 2, 3)
                # y SEGUNDO por 'llegada' (el que ingresó antes pasa primero)
                pacientes_ordenados = sorted(pacientes, key=lambda x: (x['orden'], x['llegada']))

                for i, p in enumerate(pacientes_ordenados, 1):
                    print(f"{i}. {p['nombre']} - {p['prioridad']} | Motivo: {p['motivo']}")

        elif opcion == '3':
            print("\nSaliendo del programa...")
            break
        else:
            print("Opción no válida. Intente nuevamente.")

if __name__ == "__main__":
    sistema_triage()
```

```python
from functools import reduce

class estadisticaExtra:
    def media_armonica(self, serie):
        n = len(serie)
        suma_inversos = reduce(lambda acum, num: acum + 1/float(num), serie, 0)
        return n / suma_inversos

    def frecuencia(self, serie):
        frec = {}
        for elemento in serie:
            frec[elemento] = frec.get(elemento, 0) + 1
        return frec


class palabraPalindrome:
    def esPalindrome(self, palabra):
        palabra = palabra.lower().strip()
        return palabra == palabra[::-1]


# 1) Media armónica
objEstadisticaExtra = estadisticaExtra()
serieNumeros = input("Serie de números (separados por coma): ").split(",")
print("Media armónica: ", objEstadisticaExtra.media_armonica(serieNumeros))

# 2) Frecuencia
serieDatos = input("Serie de datos para frecuencia (separados por coma): ").split(",")
print("Frecuencia: ", objEstadisticaExtra.frecuencia(serieDatos))

# 3) Palíndromo
objPalindrome = palabraPalindrome()
palabra = input("Escriba una palabra: ")
if objPalindrome.esPalindrome(palabra):
    print(palabra, "SÍ es palíndrome")
else:
    print(palabra, "NO es palíndrome")
```

```python
def calcular_impuesto(monto):

    tabla = [
        (0.01, 500.00, 1.5, 0.0),
        (500.01, 1000.00, 1.5, 3.0),
        (1000.01, 2000.00, 3.0, 3.0),
        (2000.01, 3000.00, 6.0, 3.0),
        (3000.01, 6000.00, 9.0, 2.0),
        (6000.01, 18000.00, 15.0, 2.0),
        (18000.01, 30000.00, 39.0, 2.0),
        (30000.01, 60000.00, 63.0, 1.0),
        (60000.01, 100000.00, 93.0, 0.8),
        (100000.01, 200000.00, 125.0, 0.7),
        (200000.01, 300000.00, 195.0, 0.6),
        (300000.01, 400000.00, 255.0, 0.45),
        (400000.01, 500000.00, 300.0, 0.4),
        (500000.01, 1000000.00, 340.0, 0.30),
        (1000000.01, 99999999.00, 490.0, 0.18)
    ]

    for desde, hasta, precio, adicional in tabla:
        if desde <= monto <= hasta:

            impuesto = ((monto - desde) / 1000) * adicional + precio
            return impuesto

    return 0.0

def main():
    try:
        monto = float(input("Ingrese el monto de la actividad económica ($): "))
        if monto <= 0:
            print("El monto debe ser mayor a 0.")
            return

        impuesto = calcular_impuesto(monto)
        print(f"Monto ingresado: ${monto:,.2f}")
        print(f"Impuesto a pagar: ${impuesto:,.2f}")

    except ValueError:
        print("Entrada inválida. Ingrese un valor numérico.")

if __name__ == "__main__":
    main()
```

## **2. Conversor de área**

```python



FACTORES_A_M2 = {
    'a': 0.092903,
    'b': 0.698896,
    'c': 0.836127,
    'd': 1.0,
    'e': 629.0064,
    'f': 6988.96,
    'g': 10000.0
}

NOMBRES_UNIDADES = {
    'a': 'Pie Cuadrado',
    'b': 'Vara Cuadrada',
    'c': 'Yarda Cuadrada',
    'd': 'Metro Cuadrado',
    'e': 'Tarea ',
    'f': 'Manzana ',
    'g': 'Hectárea'
}

def mostrar_menu():

    print(" CONVERSOR DE ÁREA  ")

    for opcion, nombre in NOMBRES_UNIDADES.items():
        print(f"  {opcion}. {nombre}")

def convertir_superficie(valor, origen, destino):

    metros_cuadrados = valor * FACTORES_A_M2[origen]
    return metros_cuadrados / FACTORES_A_M2[destino]

def pedir_opcion(mensaje):
    while True:
        opcion = input(mensaje).strip().lower()
        if opcion in FACTORES_A_M2:
            return opcion
        print("Opción inválida. Seleccione una letra entre 'a' y 'g'.")

def main():
    mostrar_menu()

    print("\n--- Selección de Unidades ---")
    origen = pedir_opcion("Unidad de ORIGEN (a-g): ")
    destino = pedir_opcion("Unidad de DESTINO (a-g): ")

    try:
        valor = float(input("\nIngrese la cantidad a convertir: "))
        if valor < 0:
            print("Error: La cantidad debe ser un número positivo.")
            return

        resultado = convertir_superficie(valor, origen, destino)

        print("\n================ RESULTADO ================")
        print(f"{valor:,.4f} {NOMBRES_UNIDADES[origen]} = {resultado:,.4f} {NOMBRES_UNIDADES[destino]}")

    except ValueError:
        print("Error: Debe ingresar un número válido.")

if __name__ == "__main__":
    main()
```

## **3. Sistema estadístico**

```python
from functools import reduce
from collections import Counter

class Estadistica:
    def media(self, serie):
        suma = reduce(lambda acum, num: acum + float(num), serie, 0)
        return suma / len(serie)

    def mediana(self, serie):
        datos = sorted([float(x) for x in serie])
        n = len(datos)
        mitad = n // 2
        if n % 2 == 0:
            return (datos[mitad - 1] + datos[mitad]) / 2
        return datos[mitad]

    def moda(self, serie):
        conteo = Counter(serie)
        max_frec = max(conteo.values())
        modas = [float(k) for k, v in conteo.items() if v == max_frec]
        return modas if len(modas) < len(serie) else "No hay moda"

    def varianza(self, serie):
        med = self.media(serie)
        suma = reduce(lambda acum, num: acum + (float(num) - med) ** 2, serie, 0)
        return suma / len(serie)

    def desviacion(self, serie):
        return self.varianza(serie) ** 0.5

    def rango(self, serie):
        datos = [float(x) for x in serie]
        return max(datos) - min(datos)

    def frecuencia(self, serie):
        conteo = Counter(serie)
        return dict(conteo)

def main():
    objEstadistica = Estadistica()

    entrada = input("Serie de datos (separados por coma): ")
    serie = [x.strip() for x in entrada.split(",") if x.strip()]

    if not serie:
        print("Debe ingresar al menos un dato.")
        return

    print("\n--- RESULTADOS ESTADÍSTICOS ---")
    print("a. Media aritmética:", objEstadistica.media(serie))
    print("b. Mediana:", objEstadistica.mediana(serie))
    print("c. Moda:", objEstadistica.moda(serie))
    print("d. Varianza:", objEstadistica.varianza(serie))
    print("e. Desviación estándar:", objEstadistica.desviacion(serie))
    print("f. Desviación típica:", objEstadistica.desviacion(serie))
    print("g. Rango:", objEstadistica.rango(serie))
    print("h. Frecuencia:", objEstadistica.frecuencia(serie))

    print("\n--- EXPLICACIÓN DE LA DIFERENCIA (e vs f) ---\n"
          "Diferencia: En la práctica estadística, 'Desviación Estándar' y 'Desviación Típica' "
          "son exactamente el mismo concepto y calculan el mismo valor (la raíz cuadrada de la varianza). "
          "La única diferencia es terminológica: 'Desviación Estándar' proviene de la traducción directa "
          "del inglés ('Standard Deviation'), mientras que 'Desviación Típica' es la denominación tradicional "
          "utilizada en la literatura en español.")

if __name__ == "__main__":
    main()
```

```python
from functools import reduce
from collections import Counter
import json

class SistemaAcademico:

    def __init__(self):
        self.estudiantes = []

    def agregar_estudiante(self, estudiante):
        self.estudiantes.append(estudiante)

    def mostrar_estudiantes(self):
        print("          LISTA DE ESTUDIANTES")

        for i, estudiante in enumerate(self.estudiantes, 1):
            print("\nEstudiante:", i)
            print("Nombre:", estudiante["nombre"])
            print("Materias:")
            for materia, nota in estudiante["materias"].items():
                print("  ", materia, ":", nota)
            print("Edad:", estudiante["edad"], "años")
            print(
                "Tiempo de traslado:",
                estudiante["tiempo_traslado"],
                "minutos"
            )
            print(
                "Horas de uso del telefono:",
                estudiante["horas_telefono"],
                "horas"
            )

    def guardar_json(self):
        with open(
            "estudiantes.json",
            "w",
            encoding="utf-8"
        ) as archivo:
            json.dump(
                self.estudiantes,
                archivo,
                indent=4,
                ensure_ascii=False
            )

        print("\nLos datos fueron guardados en estudiantes.json")

    def cargar_json(self):
        try:
            with open(
                "estudiantes.json",
                "r",
                encoding="utf-8"
            ) as archivo:
                self.estudiantes = json.load(archivo)
            print("\nDatos cargados correctamente.")
        except FileNotFoundError:
            print("\nNo existe el archivo estudiantes.json.")

class estadistica:

    def media(self, serie):
        suma = reduce(
            lambda acum, num:
            acum + float(num),
            serie,
            0
        )
        return suma / len(serie)
    def mediana(self, serie):
        datos = sorted(
            [float(num) for num in serie]
        )
        cantidad = len(datos)
        if cantidad % 2 == 0:
            posicion1 = cantidad // 2 - 1
            posicion2 = cantidad // 2
            return (
                datos[posicion1] +
                datos[posicion2]
            ) / 2
        else:
            return datos[cantidad // 2]
    def moda(self, serie):
        datos = [
            float(num)
            for num in serie
        ]
        frecuencias = Counter(datos)
        mayor = max(
            frecuencias.values()
        )
        if mayor == 1:
            return "No existe moda"
        modas = []
        for numero, frecuencia in frecuencias.items():
            if frecuencia == mayor:
                modas.append(numero)
        return modas
    def varianza(self, serie):
        media = self.media(serie)
        suma = reduce(
            lambda acum, num:
            acum +
            (float(num) - media) ** 2,
            serie,
            0
        )
        return suma / len(serie)
    def desviacion_estandar(self, serie):
        return self.varianza(serie) ** 0.5
    def desviacion_tipica(self, serie):
        return self.desviacion_estandar(serie)
    def rango(self, serie):
        datos = [
            float(num)
            for num in serie
        ]
        return max(datos) - min(datos)
    def frecuencia(self, serie):
        datos = [
            float(num)
            for num in serie
        ]
        return dict(
            Counter(datos)
        )
    def mostrar_todas(self, serie):
        print("       RESULTADOS ESTADISTICOS")
        print(
            "Media:",
            round(self.media(serie), 2)
        )
        print(
            "Mediana:",
            round(self.mediana(serie), 2)
        )
        print(
            "Moda:",
            self.moda(serie)
        )
        print(
            "Varianza:",
            round(self.varianza(serie), 2)
        )
        print(
            "Desviacion estandar:",
            round(
                self.desviacion_estandar(serie),
                2
            )
        )
        print(
            "Desviacion tipica:",
            round(
                self.desviacion_tipica(serie),
                2
            )
        )
        print(
            "Rango:",
            round(self.rango(serie), 2)
        )
        print(
            "Frecuencia:",
            self.frecuencia(serie)
        )

def menu_medidas(objEstadistica, serie, nombre):

    while True:
        print("     ESTADISTICAS DE", nombre)
        print("1. Media aritmetica")
        print("2. Mediana")
        print("3. Moda")
        print("4. Varianza")
        print("5. Desviacion estandar")
        print("6. Desviacion tipica")
        print("7. Rango")
        print("8. Frecuencia")
        print("9. Mostrar todas")
        print("10. Regresar")
        opcion = input("\nSeleccione una opcion: ")

        if opcion == "1":
            print(
                "\nMedia:",
                round(
                    objEstadistica.media(serie),
                    2
                )
            )
        elif opcion == "2":
            print(
                "\nMediana:",
                round(
                    objEstadistica.mediana(serie),
                    2
                )
            )
        elif opcion == "3":
            print(
                "\nModa:",
                objEstadistica.moda(serie)
            )
        elif opcion == "4":
            print(
                "\nVarianza:",
                round(
                    objEstadistica.varianza(serie),
                    2
                )
            )
        elif opcion == "5":
            print(
                "\nDesviacion estandar:",
                round(
                    objEstadistica.desviacion_estandar(serie),
                    2
                )
            )
        elif opcion == "6":
            print(
                "\nDesviacion tipica:",
                round(
                    objEstadistica.desviacion_tipica(serie),
                    2
                )
            )
        elif opcion == "7":
            print(
                "\nRango:",
                round(
                    objEstadistica.rango(serie),
                    2
                )
            )
        elif opcion == "8":
            print(
                "\nFrecuencia:",
                objEstadistica.frecuencia(serie)
            )
        elif opcion == "9":
            objEstadistica.mostrar_todas(
                serie
            )
        elif opcion == "10":
            break
        else:

            print("\nOpcion no valida.")

def menu_estadistico(
    objEstadistica,
    estudiantes
):

    while True:
        print("          SISTEMA ESTADISTICO")
        print("1. Estadisticas de notas")
        print("2. Estadisticas de edades")
        print("3. Estadisticas de tiempo de traslado")
        print("4. Estadisticas de horas de telefono")
        print("5. Regresar")

        opcion = input(
            "\nSeleccione una opcion: "
        )
        if opcion == "1":

            while True:
                print("       ESTADISTICAS DE NOTAS")
                print("1. Estadistica")
                print("2. Diseño de Circuitos Digitales")
                print("3. Ingles Basico")
                print("4. Base de Datos II")
                print("5. Programacion Computacional III")
                print("6. Regresar")
                opcion_nota = input(
                    "\nSeleccione una materia: "
                )

                materias = []
                if opcion_nota == "1":
                    materia_buscada = "Estadistica"
                elif opcion_nota == "2":
                    materia_buscada = (
                        "Diseño de Circuitos Digitales"
                    )
                elif opcion_nota == "3":
                    materia_buscada = "Ingles Basico"
                elif opcion_nota == "4":
                    materia_buscada = "Base de Datos II"
                elif opcion_nota == "5":
                    materia_buscada = (
                        "Programacion Computacional III"
                    )
                elif opcion_nota == "6":
                    break
                else:
                    print("\nOpcion no valida.")
                    continue
                # Obtener las notas
                for estudiante in estudiantes:
                    if materia_buscada in estudiante["materias"]:
                        materias.append(
                            estudiante["materias"][
                                materia_buscada
                            ]
                        )
                if len(materias) > 0:
                    menu_medidas(
                        objEstadistica,
                        materias,
                        materia_buscada
                    )
        elif opcion == "2":
            serie = []
            for estudiante in estudiantes:
                serie.append(
                    estudiante["edad"]
                )
            menu_medidas(
                objEstadistica,
                serie,
                "EDADES"
            )
        elif opcion == "3":
            serie = []
            for estudiante in estudiantes:
                serie.append(
                    estudiante[
                        "tiempo_traslado"
                    ]
                )
            menu_medidas(
                objEstadistica,
                serie,
                "TIEMPO DE TRASLADO"
            )

        elif opcion == "4":
            serie = []
            for estudiante in estudiantes:
                serie.append(
                    estudiante[
                        "horas_telefono"
                    ]
                )
            menu_medidas(
                objEstadistica,
                serie,
                "HORAS DE TELEFONO"
            )
        elif opcion == "5":
            break
        else:
            print("\nOpcion no valida.")

objAcademico = SistemaAcademico()

objEstadistica = estadistica()

objAcademico.agregar_estudiante({
    "nombre":
    "SAMUEL ALEXANDER PORTILLO CERÓN",
    "materias": {
        "Estadistica": 10,
        "Diseño de Circuitos Digitales": 10,
        "Ingles Basico": 10,
        "Base de Datos II": 10
    },
    "edad": 20,
    "tiempo_traslado": 80,
    "horas_telefono": 6
})

objAcademico.agregar_estudiante({
    "nombre":
    "CARLOS DAVID TURCIOS RIVAS",
    "materias": {
        "Estadistica": 9,
        "Diseño de Circuitos Digitales": 9,
        "Ingles Basico": 10,
        "Base de Datos II": 10
    },
    "edad": 20,
    "tiempo_traslado": 25,
    "horas_telefono": 5
})

objAcademico.agregar_estudiante({
    "nombre":
    "Gilda Mileidy González Castro",
    "materias": {
        "Estadistica": 7.5,
        "Diseño de Circuitos Digitales": 10,
        "Ingles Basico": 10,
        "Base de Datos II": 10
    },
    "edad": 19,
    "tiempo_traslado": 30,
    "horas_telefono": 1.33
})

objAcademico.agregar_estudiante({
    "nombre":
    "Kimberly Michelle Castro Zealaya",
    "materias": {
        "Programacion Computacional III": 10
    },
    "edad": 18,
    "tiempo_traslado": 20,
    "horas_telefono": 9
})

objAcademico.agregar_estudiante({
    "nombre":
    "Yeimy Julissa Parada Chávez",
    "materias": {
        "Base de Datos II": 9.5,
        "Ingles Basico": 9.5,
        "Programacion Computacional III": 9.5,
        "Diseño de Circuitos Digitales": 9.5,
        "Estadistica": 9.5
    },
    "edad": 20,
    "tiempo_traslado": 45,
    "horas_telefono": 3
})

objAcademico.agregar_estudiante({
    "nombre":
    "SAÚL ALEXANDER JIMENEZ GARCIA",
    "materias": {
        "Programacion Computacional II": 10
    },
    "edad": 21,
    "tiempo_traslado": 60,
    "horas_telefono": 5
})

objAcademico.agregar_estudiante({
    "nombre":
    "Josué Stiven Funes Benavides",
    "materias": {
        "Base de Datos II": 9.5,
        "Ingles Basico": 9.5,
        "Programacion Computacional III": 9.5,
        "Diseño de Circuitos Digitales": 9.5,
        "Estadistica": 9.5
    },
    "edad": 19,
    "tiempo_traslado": 30,
    "horas_telefono": 5
})

objAcademico.agregar_estudiante({
    "nombre":
    "JEFFERSON STEVEN CASTILLO LEIVA",
    "materias": {
        "Base de Datos II": 9.5,
        "Ingles Basico": 9.5,
        "Programacion Computacional III": 9.5,
        "Diseño de Circuitos Digitales": 9.5,
        "Estadistica": 9.5
    },
    "edad": 19,
    "tiempo_traslado": 15,
    "horas_telefono": 2.30
})

objAcademico.agregar_estudiante({
    "nombre":
    "RAMAEL ERNESTO CORDOVA COLINDRES",
    "materias": {
        "Base de Datos II": 9.5,
        "Ingles Basico": 9.5,
        "Programacion Computacional III": 9.5,
        "Diseño de Circuitos Digitales": 9.5,
        "Estadistica": 9.5
    },
    "edad": 22,
    "tiempo_traslado": 7,
    "horas_telefono": 13.52
})

objAcademico.agregar_estudiante({
    "nombre":
    "Carlos Daniel Parada Avalos",
    "materias": {
        "Base de Datos II": 8.5,
        "Ingles Basico": 8.5,
        "Programacion Computacional III": 8.5,
        "Diseño de Circuitos Digitales": 8.5,
        "Estadistica": 8.5
    },
    "edad": 25,
    "tiempo_traslado": 10,
    "horas_telefono": 5.52
})

objAcademico.agregar_estudiante({
    "nombre":
    "Karen Elizabet Ulloa Gómez",
    "materias": {
        "Base de Datos II": 9,
        "Ingles Basico": 9,
        "Programacion Computacional III": ,
        "Diseño de Circuitos Digitales": 9,
        "Estadistica": 9
    },
    "edad": 18,
    "tiempo_traslado": 40,
    "horas_telefono": 8
})

objAcademico.agregar_estudiante({
    "nombre":
    "Madison Odalis Valle Juárez",
    "materias": {
        "Base de Datos II": 9,
        "Ingles Basico": 9,
        "Programacion Computacional III": 9,
        "Diseño de Circuitos Digitales": 9,
        "Estadistica": 9
    },
    "edad": 20,
    "tiempo_traslado": 40,
    "horas_telefono": 9.33
})

objAcademico.agregar_estudiante({
    "nombre":
    "Miguel Alejandro Gómez Saravia",
    "materias": {
        "Base de Datos II": 9.5,
        "Ingles Basico": 9.5,
        "Programacion Computacional III": 9.5,
        "Diseño de Circuitos Digitales": 9.5,
        "Estadistica": 9.5
    },
    "edad": 19,
    "tiempo_traslado": 15,
    "horas_telefono": 4
})

objAcademico.agregar_estudiante({
    "nombre":"Nehemias Isaí Martínez Argueta",
    "materias": {
        "Base de Datos II": 8,
        "Ingles Basico": 8,
        "Programacion Computacional III": 8,
        "Diseño de Circuitos Digitales": 8,
        "Estadistica": 8
    },
    "edad": 19,
    "tiempo_traslado": 45,
    "horas_telefono": 9
})

objAcademico.agregar_estudiante({
    "nombre":"Sharon Nicol Argüello Torres",
    "materias": {
        "Programacion Computacional III": 8
    },
    "edad": 20,
    "tiempo_traslado": 20,
    "horas_telefono": 6
})

objAcademico.agregar_estudiante({
    "nombre":"Jenifer Maricela Tenderos Guevara",
    "materias": {
        "Base de Datos II": 9.5,
        "Ingles Basico": 9.5,
        "Programacion Computacional III": 9.5,
        "Diseño de Circuitos Digitales": 9.5,
        "Estadistica": 9.5
    },
    "edad": 21,
    "tiempo_traslado": 60,
    "horas_telefono": 3
})

objAcademico.agregar_estudiante({
    "nombre":"Claudia Arely Sorto Meléndez",
    "materias": {
        "Base de Datos II": 8.5,
        "Ingles Basico": 8.5,
        "Programacion Computacional III": 8.5,
        "Diseño de Circuitos Digitales": 8.5,
        "Estadistica": 8.5
    },
    "edad": 19,
    "tiempo_traslado": 50,
    "horas_telefono": 7.62
})


objAcademico.agregar_estudiante({
    "nombre":"Daniel Enrique Quintanilla Paniagua",
    "materias": {
        "Base de Datos II": 9.5,
        "Ingles Basico": 9.5,
        "Programacion Computacional III": 9.5,
        "Diseño de Circuitos Digitales": 9.5,
        "Estadistica": 9.5
    },
    "edad": 19,
    "tiempo_traslado": 15,
    "horas_telefono": 4
})


objAcademico.agregar_estudiante({
    "nombre": "José Roberto Díaz",
    "materias": {
        "Base de Datos II": 9,
        "Ingles Basico": 9,
        "Programacion Computacional III": 9,
        "Diseño de Circuitos Digitales": 9,
        "Estadistica": 9
    },
    "edad": 21,
    "tiempo_traslado": 75,
    "horas_telefono": 5
})

objAcademico.agregar_estudiante({
    "nombre": "Kevin Manrique Campos",
    "materias": {
        "Base de Datos II": 9.5,
        "Ingles Basico": 9.5,
        "Programacion Computacional III": 9.5,
        "Diseño de Circuitos Digitales": 9.5,
        "Estadistica": 9.5
    },
    "edad": 19,
    "tiempo_traslado": 30,
    "horas_telefono": 6
})

while True:
    print("\n")
    print("         SISTEMA ACADEMICO UGB")
    print("1. Mostrar estudiantes")
    print("2. Guardar datos en JSON")
    print("3. Cargar datos desde JSON")
    print("4. Sistema estadistico")
    print("5. Salir")

    opcion = input(
        "\nSeleccione una opcion: "
    )

    if opcion == "1":
        objAcademico.mostrar_estudiantes()
    elif opcion == "2":
        objAcademico.guardar_json()
    elif opcion == "3":
        objAcademico.cargar_json()
    elif opcion == "4":

        menu_estadistico(
            objEstadistica,
            objAcademico.estudiantes
        )

    elif opcion == "5":
        print("\nPrograma finalizado.")
        break
    else:
        print("\nOpcion no valida.")
```
