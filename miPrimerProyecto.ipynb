Funciones en Python

```python
def suma(num1, num2):
  respuesta = float(num1) + float(num2)
  return respuesta

num1 = input("Num 1: ")
num2 = input("Num 2: ")
print("Respesta ", suma(num1, num2))
```

```python
def resta(num1, num2):
  respuesta = float(num1) - float(num2)
  return respuesta

num1 = input("Num 1: ")
num2 = input("Num 2: ")
print("Respesta ", resta(num1, num2))
```

De grados  Fahrenheit a Celsius.
Fórmula Fahrenheit a Celsius: C = (F - 32) * 5/9

```python
def convertir(fahrenheit):
    return (float(fahrenheit) - 32) * 5/9

f = input("Grados Fahrenheit: ")
print("Grados Celsius:", convertir(f))
```

De grados Celsius a Fahrenheit.
Fórmula Celsius a Fahrenheit:
F = (C * 9/5) + 32:


```python
def convertir(celsius):
  # Fórmula: (C * 9/5) + 32
    return float(celsius) * 9/5 + 32

c = input("Grados Celsius: ")
print("Grados Fahrenheit:", convertir(c))
```

De grados Celsius a Kelvin. Fórmula Celsius a Kelvin: K = C + 273.15

```python
def convertir(celsius):
    return float(celsius) + 273.15

c = input("Grados Celsius: ")
print("Grados Kelvin:", convertir(c))
```

Funciones anonimas (No resiven nombre)

```python
convertir= lambda c: float(c) *9/5 + 32

c = float(input("Grados Celsius: "))
print("Grados Fahrenheit:", convertir(c))
```
