```python
print("Sumar dos numeros")
num1 = input("num 1: ")
num2 = input("num 2: ")
respuesta = float(num1) + float(num2)

print(type(num1), type(num2), type(respuesta))

print("Respuesta: ", respuesta)

respuesta = "Suma completa"
print(respuesta, type(respuesta))
```

```python
print("Restar dos numeros")
num1 = input("num 1: ")
num2 = input("num 2: ")
respuesta = float(num1) - float(num2)

print(type(num1), type(num2), type(respuesta))

print("Respuesta: ", respuesta)

respuesta = "Resta completa"
print(respuesta, type(respuesta))
```

```python
print("Multiplicar dos numeros")
num1 = input("num 1: ")
num2 = input("num 2: ")
respuesta = float(num1) * float(num2)

print(type(num1), type(num2), type(respuesta))

print("Respuesta: ", respuesta)

respuesta = "Multiplicacion completa"
print(respuesta, type(respuesta))
```

```python
print("Dividir dos numeros")
num1 = input("num 1: ")
num2 = input("num 2: ")
respuesta = float(num1) / float(num2)

print(type(num1), type(num2), type(respuesta))

print("Respuesta: ", respuesta)

respuesta = "Dividir completa"
print(respuesta, type(respuesta))
```
