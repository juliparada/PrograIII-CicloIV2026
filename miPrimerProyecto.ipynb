Calculo de sueldo de un empleado con deducciones: Sueldo base = horas * valor_hora * 30
Deducciones: ISSS (3%), AFP (7.25%), ISR según tabla MH

```python
def calcular(horas, valor_hora):
    sueldo_base = float(horas) * float(valor_hora) * 30

    # Deducciones
    isss = sueldo_base * 0.03
    afp = sueldo_base * 0.0725
    sueldo_gravado = sueldo_base - isss - afp

    # ISR según tabla MH
    if sueldo_gravado <= 550:
        isr = 0
    elif sueldo_gravado <= 895.24:
        isr = (sueldo_gravado - 550) * 0.10 + 17.67
    elif sueldo_gravado <= 2038.10:
        isr = (sueldo_gravado - 895.24) * 0.20 + 60.00
    else:
        isr = (sueldo_gravado - 2038.10) * 0.30 + 288.57

    sueldo_neto = sueldo_gravado - isr
    return sueldo_base, sueldo_neto, isss, afp, isr

h = input("Horas trabajadas por día: ")
v = input("Valor de la hora: ")
base, neto, isss, afp, isr = calcular(h, v)

print("Sueldo base:", base)
print("ISSS:", isss)
print("AFP:", afp)
print("ISR:", isr)
print("Sueldo total:", neto)
```

1.Calcular sobre el sueldo las deducciones por ley ISSS (3%), AFP(6.25%), ISR(Según tabla) e imprima el total deducciones y sueldo neto.

```python
while True:
    sueldo = float(input("Ingrese el sueldo: $"))

    if sueldo * 0.03 > 30.00:
        isss = 30.00
    else:
        isss = sueldo * 0.03

    if sueldo * 0.0725 > 507.50:
        afp = 507.50
    else:
        afp = sueldo * 0.0725

    renta_gravada = sueldo - isss - afp

    if renta_gravada <= 550.00:
        isr = 0.00
    elif renta_gravada <= 895.24:
        isr = (17.67 + (renta_gravada - 550.00) * 0.10)
    elif renta_gravada <= 2038.10:
        isr = (60.00 + (renta_gravada - 895.24) * 0.20)
    else:
        isr = (288.57 + (renta_gravada - 2038.10) * 0.30)

    total_deducciones = isss + afp + isr

    sueldo_neto = sueldo - total_deducciones

    print("\n--- DEDUCCIONES ---")
    print(f"Sueldo:             ${sueldo:.2f}")
    print(f"ISSS (3%):          ${isss:.2f}")
    print(f"AFP (7.25%):        ${afp:.2f}")
    print(f"ISR:                ${isr:.2f}")
    print(f"Total deducciones:  ${total_deducciones:.2f}")
    print(f"Sueldo neto:        ${sueldo_neto:.2f}")

    continuar = input("\n¿Desea calcular otro sueldo? (s/n): ").lower()
    if continuar != 's':
        break
print("Fin del cálculo de sueldo.")
```

6.Nivel de prioridad de pacientes

```python
import datetime

def evaluar_prioridad(edad, motivo):
    motivo_lower = motivo.lower()

    # PRIORIDAD 1: Emergencias vitales o signos de gravedad agudos
    rojo_directo = [
        "inconsciente", "paro", "shock", "choque", "hemorragia masiva",
        "politraumatismo", "no respira", "convulsion", "paro cardiaco", "infarto"
    ]
    if any(kw in motivo_lower for kw in rojo_directo):
        return "Prioridad 1"

    sospecha_emergencia = ["dolor de pecho", "dificultad para respirar", "falta de aire", "asfixia"]
    if any(kw in motivo_lower for kw in sospecha_emergencia):
        resp = input("  [?] ¿El paciente presenta dificultad respiratoria severa o dolor de pecho agudo? (s/n): ").strip().lower()
        if resp == 's':
            return "Prioridad 1"

    if edad < 5 and any(kw in motivo_lower for kw in ["fiebre", "tos", "diarrea", "vomito", "grave"]):
        resp = input("  [?] ¿El menor presenta letargia, convulsiones o incapacidad para beber/beber pecho? (s/n): ").strip().lower()
        if resp == 's':
            return "Prioridad 1"

    if any(kw in motivo_lower for kw in ["embarazo", "embarazada", "parto", "puerpera", "sangrado vaginal"]):
        resp = input("  [?] ¿Presenta sangrado vaginal, convulsiones o fuerte dolor abdominal? (s/n): ").strip().lower()
        if resp == 's':
            return "Prioridad 1"

    # PRIORIDAD 2: Vulnerables o casos agudos/descompensados
    if edad >= 65 or edad < 1:
        return "Prioridad 2"

    if any(kw in motivo_lower for kw in ["embarazo", "embarazada", "puerpera"]):
        return "Prioridad 2"

    p2_keywords = [
        "descompensad", "diabetes", "hipertension", "presion alta", "presion baja",
        "dolor fuerte", "dolor moderado", "fractura", "quemadura", "alergia", "asma"
    ]
    if any(kw in motivo_lower for kw in p2_keywords):
        return "Prioridad 2"

    # PRIORIDAD 3: Consultas leves o rutinarias
    return "Prioridad 3"


def sistema_triage():
    pacientes = []
    orden_prioridad = {"Prioridad 1": 1, "Prioridad 2": 2, "Prioridad 3": 3}
    contador_llegada = 0  # Garantiza el orden de llegada exacto

    while True:
        print("\n--- MENÚ PRINCIPAL ---")
        print("1. Agregar paciente")
        print("2. Ver lista de pacientes por orden de atención")
        print("3. Salir")
        opcion = input("Seleccione una opción: ").strip()

        if opcion == '1':
            print("\n--- NUEVO PACIENTE ---")
            nombre = input("Nombre del paciente: ").strip()

            while True:
                try:
                    edad = int(input("Edad: "))
                    if edad >= 0:
                        break
                    print("Ingrese una edad válida.")
                except ValueError:
                    print("Por favor, ingrese un número entero.")

            motivo = input("Motivo de consulta: ").strip()

            prioridad = evaluar_prioridad(edad, motivo)
            contador_llegada += 1  # Asigna turno correlativo según llegada

            pacientes.append({
                "nombre": nombre,
                "motivo": motivo,
                "prioridad": prioridad,
                "orden": orden_prioridad[prioridad],
                "llegada": contador_llegada
            })

            print(f"\n[OK] Paciente {nombre.upper()} asignado a: {prioridad}")

            sub_opcion = input("\n¿Desea agregar otro paciente? (s/n): ").strip().lower()
            if sub_opcion != 's':
                pass

        elif opcion == '2':
            print("\n--- LISTA DE PACIENTES (ORDEN DE ATENCIÓN) ---")
            if not pacientes:
                print("No hay pacientes registrados.")
            else:
                # Se ordena PRIMERO por 'orden' (Prioridad 1, 2, 3)
                # y SEGUNDO por 'llegada' (el que ingresó antes pasa primero)
                pacientes_ordenados = sorted(pacientes, key=lambda x: (x['orden'], x['llegada']))

                for i, p in enumerate(pacientes_ordenados, 1):
                    print(f"{i}. {p['nombre']} - {p['prioridad']} | Motivo: {p['motivo']}")

        elif opcion == '3':
            print("\nSaliendo del programa...")
            break
        else:
            print("Opción no válida. Intente nuevamente.")

if __name__ == "__main__":
    sistema_triage()
```

```python
from functools import reduce
class estadistica:
  def media(self, serie):
    suma = reduce(lambda acum, num: acum+ float(num), serie, 0)
    return suma/len(serie)

  def varianza(self, serie):
    media = self.media(serie)
    #4 + 0 + 4 + 0 + 0 + 16= 24/6 = 4.0
    suma = reduce(lambda acum, num: acum + (float(num)-media)**2, serie, 0)
    return suma/len(serie)

  def desviacion(self, serie):
    return self.varianza(serie)**0.5
```

```python
import json

class Estudiante:
    def __init__(self, nombre, edad, tiempo_traslado, horas_telefono):
        self.nombre = nombre
        self.edad = edad
        self.tiempo_traslado = tiempo_traslado
        self.horas_telefono = horas_telefono
        # Diccionario de materias con sus notas
        self.materias = {
            "Bases de Datos II": None,
            "Inglés Básico": None,
            "Programación Computacional III": None,
            "Diseño de Circuitos Digitales": None,
            "Estadística Computacional": None
        }

    def to_dict(self):
        return {
            "nombre": self.nombre,
            "edad": self.edad,
            "tiempo_traslado": self.tiempo_traslado,
            "horas_telefono": self.horas_telefono,
            "materias": self.materias
        }

    def __str__(self):
        return (f"{self.nombre:35} | Edad: {self.edad:3} | "
                f"Traslado: {self.tiempo_traslado:3} min | "
                f"Cel: {self.horas_telefono:5} h")


class SistemaEstadistico:
    def __init__(self):
        self.estudiantes = []

    def cargar_datos_ejemplo(self):
        # Datos de ejemplo con notas reales
        datos_ejemplo = [
          ("Carlos David Turcios Rivas", 20, 25, 5.0, [8, 7, 9, 7, 8]),
          ("Gilda Mileidy González Castro", 19, 30, 1.5, [10, 10, 9, 10, 9]),
          ("Samuel Alexander Portillo Cerón", 20, 80, 6.0, [8, 7, 8, 8, 7]),
          ("Kimberly Michelle Castro Zelaya", 18, 20, 9.0, [10, 8, 10, 9, 8]),
          ("Yeimy Julissa Parada Chavez", 20, 45, 3.0, [10, 9, 9, 8, 9]),
          ("Saul Alexander Jimenez Garcia", 21, 60, 5.0, [9, 8, 10, 9, 8]),
          ("Miguel Alejandro Gómez Saravia", 19, 15, 4.0, [9, 10, 8, 8, 9]),
          ("Josué Stiven Funes Benavides", 19, 30, 5.0, [9, 8, 10, 9, 8]),
          ("Jefferson Steven Castillo Leiva", 19, 15, 2.3, [8, 9, 10, 9, 8]),
          ("Karen Elizabet Ulloa Gómez", 18, 40, 8.0, [8, 9, 8, 7, 7]),
          ("Madison Odalis Valle Juárez", 20, 40, 9.3, [9, 8, 9, 9, 8]),
          ("Carlos Daniel Parada Avalos", 25, 10, 5.5, [8, 8, 9, 8, 8]),
          ("Nehemias Isaí Martínez Argüello", 19, 45, 9.0, [7, 8, 6, 7, 6]),
          ("Sharon Nicol Argüello Torres", 20, 20, 6.0, [8, 7, 9, 8, 7]),
          ("Ramael Ernesto Córdova Colindres", 22, 7, 13.5, [10, 9, 9, 10, 9]),
          ("Jenifer Maricela Tenderos Guevara", 21, 60, 3.0, [9, 10, 9, 9, 10]),
          ("Claudia Arely Sorto Meléndez", 19, 50, 7.6, [8, 7, 7, 8, 8]),
          ("Daniel Enrique Quintanilla Paniagua", 19, 15, 4.0, [9, 10, 8, 8, 9]),
          ("José Roberto Díaz", 21, 75, 5.0, [8, 8, 8, 8, 9]),
          ("Kevin Manrique Campos", 19, 30, 6.0, [10, 9, 9, 8, 9])
        ]

        # Get the list of subject names from a temporary Estudiante object
        temp_estudiante = Estudiante("", 0, 0, 0)
        materias_lista = list(temp_estudiante.materias.keys())

        for datos in datos_ejemplo:
            nombre, edad, traslado, horas, notas = datos
            estudiante = Estudiante(nombre, edad, traslado, horas)
            for i, materia_nombre in enumerate(materias_lista):
                if i < len(notas):
                    estudiante.materias[materia_nombre] = notas[i]
            self.estudiantes.append(estudiante)

        print(f"\n Se cargaron {len(self.estudiantes)} estudiantes de ejemplo.")


    def obtener_materias(self):
        if self.estudiantes:
            return self.estudiantes[0].materias
        # If no students, return the default subject structure from a new Estudiante
        return Estudiante("",0,0,0).materias

    def obtener_valores(self, campo, materia=None):
        valores = []
        for est in self.estudiantes:
            if campo == "edad":
                valores.append(est.edad)
            elif campo == "tiempo_traslado":
                valores.append(est.tiempo_traslado)
            elif campo == "horas_telefono":
                valores.append(est.horas_telefono)
            elif campo == "materia" and materia:
                valores.append(est.materias.get(materia))
            elif campo == "todas_notas":
                valores.extend([n for n in est.materias.values() if n is not None])
        return valores

    # FUNCIONES ESTADÍSTICAS
    @staticmethod
    def media(valores):
        valores_limpios = [v for v in valores if v is not None]
        if not valores_limpios:
            return 0
        return sum(valores_limpios) / len(valores_limpios)

    @staticmethod
    def mediana(valores):
        valores_limpios = sorted([v for v in valores if v is not None])
        if not valores_limpios:
            return 0
        n = len(valores_limpios)
        mitad = n // 2
        if n % 2 == 0:
            return (valores_limpios[mitad - 1] + valores_limpios[mitad]) / 2
        else:
            return valores_limpios[mitad]

    @staticmethod
    def moda(valores):
        valores_limpios = [v for v in valores if v is not None]
        if not valores_limpios:
            return "No hay datos"
        conteo = {}
        for v in valores_limpios:
            conteo[v] = conteo.get(v, 0) + 1
        max_repeticiones = max(conteo.values())
        modas = [valor for valor, veces in conteo.items() if veces == max_repeticiones]
        if len(modas) == len(conteo) or max_repeticiones == 1:
            return "No hay moda (todos se repiten igual)"
        return modas

    @staticmethod
    def varianza(valores):
        valores_limpios = [v for v in valores if v is not None]
        if not valores_limpios:
            return 0
        m = SistemaEstadistico.media(valores_limpios)
        return sum((v - m) ** 2 for v in valores_limpios) / len(valores_limpios)

    @staticmethod
    def desviacion_estandar(valores):
        return SistemaEstadistico.varianza(valores) ** 0.5

    @staticmethod
    def desviacion_tipica(valores):
        return SistemaEstadistico.desviacion_estandar(valores)

    @staticmethod
    def rango(valores):
        valores_limpios = [v for v in valores if v is not None]
        if not valores_limpios:
            return 0
        return max(valores_limpios) - min(valores_limpios)

    @staticmethod
    def frecuencia(valores):
        valores_limpios = [v for v in valores if v is not None]
        if not valores_limpios:
            return {}
        conteo = {}
        for v in valores_limpios:
            conteo[v] = conteo.get(v, 0) + 1
        total = len(valores_limpios)
        tabla = {}
        for valor, veces in conteo.items():
            porcentaje = round((veces / total) * 100, 1)
            tabla[valor] = {"frecuencia_absoluta": veces, "frecuencia_relativa_%": porcentaje}
        return tabla

    def guardar_json(self, ruta="estudiantes.json"):
        lista_dict = [est.to_dict() for est in self.estudiantes]
        with open(ruta, "w", encoding="utf-8") as archivo:
            json.dump(lista_dict, archivo, indent=4, ensure_ascii=False)
        print(f"\n Datos guardados en: {ruta}")

    def cargar_json(self, ruta="estudiantes.json"):
        try:
            with open(ruta, "r", encoding="utf-8") as archivo:
                lista_dict = json.load(archivo)
            self.estudiantes = []
            for d in lista_dict:
                est = Estudiante(d["nombre"], d["edad"], d["tiempo_traslado"], d["horas_telefono"])
                est.materias = d.get("materias", {})
                self.estudiantes.append(est)
            print(f"\n Se cargaron {len(self.estudiantes)} estudiantes desde {ruta}")
        except FileNotFoundError:
            print("\n No se encontró ese archivo.")

    def mostrar_estudiantes(self):
        print("LISTA DE ESTUDIANTES")
        for est in self.estudiantes:
            print(est)
            print(f"   Notas: {', '.join([f'{k}: {v if v is not None else 'Sin nota'}' for k, v in est.materias.items()])}")


def menu_estadisticas(sistema, campo, nombre_bonito, es_numerico=True):
    """Menú para calcular estadísticas de un conjunto de datos"""
    while True:
        print(f"ANÁLISIS ESTADÍSTICO - {nombre_bonito}")
        print("1. Media")
        print("2. Mediana")
        print("3. Moda")
        print("4. Varianza")
        print("5. Desviación Estándar")
        print("6. Desviación Típica")
        print("7. Rango")
        print("8. Frecuencia")
        print("9. Calcular TODAS las estadísticas")
        print("0. Regresar")

        opcion = input("\nElige una opción: ")

        if opcion == "0":
            return

        # Obtener valores según el campo
        if campo == "materia":
            valores = sistema.obtener_valores("materia", nombre_bonito)
        else:
            valores = sistema.obtener_valores(campo)

        # Verificar si hay datos
        valores_limpios = [v for v in valores if v is not None]
        if not valores_limpios:
            print("\n No hay datos disponibles para esta categoría.")
            continue

        # Mostrar resultados
        if opcion == "1":
            print(f"\n Media: {SistemaEstadistico.media(valores_limpios):.2f}")
        elif opcion == "2":
            print(f"\n Mediana: {SistemaEstadistico.mediana(valores_limpios):.2f}")
        elif opcion == "3":
            moda = SistemaEstadistico.moda(valores_limpios)
            print(f"\n Moda: {moda}")
        elif opcion == "4":
            print(f"\n Varianza: {SistemaEstadistico.varianza(valores_limpios):.2f}")
        elif opcion == "5":
            print(f"\n Desviación Estándar: {SistemaEstadistico.desviacion_estandar(valores_limpios):.2f}")
        elif opcion == "6":
            print(f"\n Desviación Típica: {SistemaEstadistico.desviacion_tipica(valores_limpios):.2f}")
        elif opcion == "7":
            print(f"\n Rango: {SistemaEstadistico.rango(valores_limpios):.2f}")
        elif opcion == "8":
            tabla = SistemaEstadistico.frecuencia(valores_limpios)
            print(f"\n{'Valor':30} | {'Frec. Absoluta':15} | {'Frec. Relativa (%)'}")
            for valor, datos in tabla.items():
                print(f"{str(valor):30} | {datos['frecuencia_absoluta']:15} | {datos['frecuencia_relativa_%']}")
        elif opcion == "9":
            print(f"\nMedia: {SistemaEstadistico.media(valores_limpios):.2f}")
            print(f"Mediana: {SistemaEstadistico.mediana(valores_limpios):.2f}")
            print(f"Moda: {SistemaEstadistico.moda(valores_limpios)}")
            print(f"Varianza: {SistemaEstadistico.varianza(valores_limpios):.2f}")
            print(f"Desviación Estándar: {SistemaEstadistico.desviacion_estandar(valores_limpios):.2f}")
            print(f"Desviación Típica: {SistemaEstadistico.desviacion_tipica(valores_limpios):.2f}")
            print(f"Rango: {SistemaEstadistico.rango(valores_limpios):.2f}")
            tabla = SistemaEstadistico.frecuencia(valores_limpios)
            print(f"\n{'Valor':30} | {'Frec. Absoluta':15} | {'Frec. Relativa (%)'}")
            for valor, datos in tabla.items():
                print(f"{str(valor):30} | {datos['frecuencia_absoluta']:15} | {datos['frecuencia_relativa_%']}")
        else:
            print("\n Opción inválida.")


def menu_principal():
    sistema = SistemaEstadistico()

    while True:
        #print("\n" + "="*70)
        print("SISTEMA ESTADÍSTICO - Programación Computacional III")
        #print("="*70)
        print("1. Cargar datos de ejemplo")
        print("2. Mostrar estudiantes")
        print("3. Calcular estadísticas por categoría")
        print("4. Guardar datos en JSON")
        print("5. Cargar datos desde JSON")
        print("6. Explicar diferencias entre medidas")
        print("0. Salir")

        opcion = input("\nElige una opción: ")

        if opcion == "0":
            print("\n Saliendo del sistema...")
            break
        elif opcion == "1":
            sistema.cargar_datos_ejemplo()
        elif opcion == "2":
            if not sistema.estudiantes:
                print("\n Primero carga datos (opción 1 o 5).")
            else:
                sistema.mostrar_estudiantes()
        elif opcion == "3":
            if not sistema.estudiantes:
                print("\n Primero carga datos (opción 1 o 5).")
            else:
                # Menú para seleccionar qué datos analizar
                while True:
                    print(" SELECCIONA QUÉ DATOS QUIERES ANALIZAR")
                    print("1. Notas de una materia específica")
                    print("2. Todas las notas (todas las materias combinadas)")
                    print("3. Edades")
                    print("4. Tiempo de traslado a la UGB")
                    print("5. Horas de uso del teléfono")
                    print("6. Mostrar materias disponibles")
                    print("0. Regresar")

                    sub_opcion = input("\nElige una opción: ")

                    if sub_opcion == "0":
                        break
                    elif sub_opcion == "1":
                        # Mostrar materias disponibles
                        materias = sistema.obtener_materias()
                        if not materias:
                            print("\n No hay materias disponibles.")
                            continue

                        print("\n MATERIAS DISPONIBLES:")
                        for i, materia in enumerate(materias.keys(), 1):
                            print(f"{i}. {materia}")

                        try:
                            seleccion = input("\n Elige el número de la materia: ")
                            idx = int(seleccion) - 1
                            materias_lista = list(materias.keys())
                            if 0 <= idx < len(materias_lista):
                                materia_seleccionada = materias_lista[idx]
                                menu_estadisticas(sistema, "materia", materia_seleccionada, True)
                            else:
                                print("\n Opción inválida.")
                        except ValueError:
                            print("\n Ingresa un número válido.")

                    elif sub_opcion == "2":
                        menu_estadisticas(sistema, "todas_notas", "Todas las Notas", True)

                    elif sub_opcion == "3":
                        menu_estadisticas(sistema, "edad", "Edades", True)

                    elif sub_opcion == "4":
                        menu_estadisticas(sistema, "tiempo_traslado", "Tiempo de Traslado (min)", True)

                    elif sub_opcion == "5":
                        menu_estadisticas(sistema, "horas_telefono", "Horas de Uso del Teléfono", True)

                    elif sub_opcion == "6":
                        materias = sistema.obtener_materias()
                        if materias:
                            print("\n MATERIAS DISPONIBLES:")
                            for i, materia in enumerate(materias.keys(), 1):
                                print(f"   {i}. {materia}")
                        else:
                            print("\n No hay materias disponibles.")

                    else:
                        print("\n Opción inválida.")

        elif opcion == "4":
            if not sistema.estudiantes:
                print("\n No hay datos para guardar.")
            else:
                sistema.guardar_json()
        elif opcion == "5":
            sistema.cargar_json()
        elif opcion == "6":
            print("EXPLICACIÓN DE MEDIDAS ESTADÍSTICAS")
            print("""
MEDIDAS DE TENDENCIA CENTRAL:
   • Media: El promedio de todos los valores.
   • Mediana: El valor que está en el centro cuando ordenamos los datos.
   • Moda: El valor que más se repite.
MEDIDAS DE DISPERSIÓN:
   • Varianza: Mide qué tan lejos están los valores de la media.
   • Desviación Estándar: Raíz cuadrada de la varianza. Es más fácil de interpretar porque está en la misma unidad que los datos.
   • Desviación Típica: Es lo mismo que la desviación estándar.
   • Rango: La diferencia entre el valor máximo y el mínimo.
FRECUENCIA:
   • Muestra cuántas veces aparece cada valor y su porcentaje.
            """)
        else:
            print("\n Opción inválida.")


if __name__ == "__main__":
    menu_principal()
```
