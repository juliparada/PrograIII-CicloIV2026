Calculo de sueldo de un empleado con deducciones: Sueldo base = horas * valor_hora * 30
Deducciones: ISSS (3%), AFP (7.25%), ISR según tabla MH

```python
def calcular(horas, valor_hora):
    sueldo_base = float(horas) * float(valor_hora) * 30

    # Deducciones
    isss = sueldo_base * 0.03
    afp = sueldo_base * 0.0725
    sueldo_gravado = sueldo_base - isss - afp

    # ISR según tabla MH
    if sueldo_gravado <= 550:
        isr = 0
    elif sueldo_gravado <= 895.24:
        isr = (sueldo_gravado - 550) * 0.10 + 17.67
    elif sueldo_gravado <= 2038.10:
        isr = (sueldo_gravado - 895.24) * 0.20 + 60.00
    else:
        isr = (sueldo_gravado - 2038.10) * 0.30 + 288.57

    sueldo_neto = sueldo_gravado - isr
    return sueldo_base, sueldo_neto, isss, afp, isr

h = input("Horas trabajadas por día: ")
v = input("Valor de la hora: ")
base, neto, isss, afp, isr = calcular(h, v)

print("Sueldo base:", base)
print("ISSS:", isss)
print("AFP:", afp)
print("ISR:", isr)
print("Sueldo total:", neto)
```

1.Calcular sobre el sueldo las deducciones por ley ISSS (3%), AFP(6.25%), ISR(Según tabla) e imprima el total deducciones y sueldo neto.

```python
while True:
    sueldo = float(input("Ingrese el sueldo: $"))

    # Cálculo de ISSS (3% con tope de $30.00)
    if sueldo * 0.03 > 30.00:
        isss = 30.00
    else:
        isss = sueldo * 0.03

    # Cálculo de AFP (7.25% con tope de $507.50)
    if sueldo * 0.0725 > 507.50:
        afp = 507.50
    else:
        afp = sueldo * 0.0725

    # Calcular la Renta Gravada
    renta_gravada = sueldo - isss - afp

    # Cálculo de ISR usando if/elif/else
    if renta_gravada <= 550.00:
        isr = 0.00
    elif renta_gravada <= 895.24:
        isr = (17.67 + (renta_gravada - 550.00) * 0.10)
    elif renta_gravada <= 2038.10:
        isr = (60.00 + (renta_gravada - 895.24) * 0.20)
    else:
        isr = (288.57 + (renta_gravada - 2038.10) * 0.30)

    total_deducciones = isss + afp + isr

    # Sueldo neto
    sueldo_neto = sueldo - total_deducciones

    print("\n--- DEDUCCIONES ---")
    print(f"Sueldo:             ${sueldo:.2f}")
    print(f"ISSS (3%):          ${isss:.2f}")
    print(f"AFP (7.25%):        ${afp:.2f}")
    print(f"ISR:                ${isr:.2f}")
    print(f"Total deducciones:  ${total_deducciones:.2f}")
    print(f"Sueldo neto:        ${sueldo_neto:.2f}")

    continuar = input("\n¿Desea calcular otro sueldo? (s/n): ").lower()
    if continuar != 's':
        break
print("Fin del cálculo de sueldo.")
```

6.Nivel de prioridad de pacientes

```python
import datetime

def evaluar_prioridad(edad, motivo):
    motivo_lower = motivo.lower()

    # PRIORIDAD 1: Emergencias vitales o signos de gravedad agudos
    rojo_directo = [
        "inconsciente", "paro", "shock", "choque", "hemorragia masiva",
        "politraumatismo", "no respira", "convulsion", "paro cardiaco", "infarto"
    ]
    if any(kw in motivo_lower for kw in rojo_directo):
        return "Prioridad 1"

    sospecha_emergencia = ["dolor de pecho", "dificultad para respirar", "falta de aire", "asfixia"]
    if any(kw in motivo_lower for kw in sospecha_emergencia):
        resp = input("  [?] ¿El paciente presenta dificultad respiratoria severa o dolor de pecho agudo? (s/n): ").strip().lower()
        if resp == 's':
            return "Prioridad 1"

    if edad < 5 and any(kw in motivo_lower for kw in ["fiebre", "tos", "diarrea", "vomito", "grave"]):
        resp = input("  [?] ¿El menor presenta letargia, convulsiones o incapacidad para beber/beber pecho? (s/n): ").strip().lower()
        if resp == 's':
            return "Prioridad 1"

    if any(kw in motivo_lower for kw in ["embarazo", "embarazada", "parto", "puerpera", "sangrado vaginal"]):
        resp = input("  [?] ¿Presenta sangrado vaginal, convulsiones o fuerte dolor abdominal? (s/n): ").strip().lower()
        if resp == 's':
            return "Prioridad 1"

    # PRIORIDAD 2: Vulnerables o casos agudos/descompensados
    if edad >= 65 or edad < 1:
        return "Prioridad 2"

    if any(kw in motivo_lower for kw in ["embarazo", "embarazada", "puerpera"]):
        return "Prioridad 2"

    p2_keywords = [
        "descompensad", "diabetes", "hipertension", "presion alta", "presion baja",
        "dolor fuerte", "dolor moderado", "fractura", "quemadura", "alergia", "asma"
    ]
    if any(kw in motivo_lower for kw in p2_keywords):
        return "Prioridad 2"

    # PRIORIDAD 3: Consultas leves o rutinarias
    return "Prioridad 3"


def sistema_triage():
    pacientes = []
    orden_prioridad = {"Prioridad 1": 1, "Prioridad 2": 2, "Prioridad 3": 3}
    contador_llegada = 0  # Garantiza el orden de llegada exacto

    while True:
        print("\n--- MENÚ PRINCIPAL ---")
        print("1. Agregar paciente")
        print("2. Ver lista de pacientes por orden de atención")
        print("3. Salir")
        opcion = input("Seleccione una opción: ").strip()

        if opcion == '1':
            print("\n--- NUEVO PACIENTE ---")
            nombre = input("Nombre del paciente: ").strip()

            while True:
                try:
                    edad = int(input("Edad: "))
                    if edad >= 0:
                        break
                    print("Ingrese una edad válida.")
                except ValueError:
                    print("Por favor, ingrese un número entero.")

            motivo = input("Motivo de consulta: ").strip()

            prioridad = evaluar_prioridad(edad, motivo)
            contador_llegada += 1  # Asigna turno correlativo según llegada

            pacientes.append({
                "nombre": nombre,
                "motivo": motivo,
                "prioridad": prioridad,
                "orden": orden_prioridad[prioridad],
                "llegada": contador_llegada
            })

            print(f"\n[OK] Paciente {nombre.upper()} asignado a: {prioridad}")

            sub_opcion = input("\n¿Desea agregar otro paciente? (s/n): ").strip().lower()
            if sub_opcion != 's':
                pass

        elif opcion == '2':
            print("\n--- LISTA DE PACIENTES (ORDEN DE ATENCIÓN) ---")
            if not pacientes:
                print("No hay pacientes registrados.")
            else:
                # Se ordena PRIMERO por 'orden' (Prioridad 1, 2, 3)
                # y SEGUNDO por 'llegada' (el que ingresó antes pasa primero)
                pacientes_ordenados = sorted(pacientes, key=lambda x: (x['orden'], x['llegada']))

                for i, p in enumerate(pacientes_ordenados, 1):
                    print(f"{i}. {p['nombre']} - {p['prioridad']} | Motivo: {p['motivo']}")

        elif opcion == '3':
            print("\nSaliendo del programa...")
            break
        else:
            print("Opción no válida. Intente nuevamente.")

if __name__ == "__main__":
    sistema_triage()
```
